```python
# Drop useless columns
df = df.drop(columns=['url', 'address', 'phone', 'menu_item', 'reviews_list', 'dish_liked'], errors='ignore')
```

```python
df['rate'] = df['rate'].astype(str).str.split('/').str[0].str.strip()
df['rate'] = df['rate'].replace(['NEW', '-', 'nan'], None)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
```

```python
df.drop_duplicates()
```

```python
df
```

```python
df['online_order'].unique()
```

```python
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].astype(str).str.replace(',', '')
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')
df = df.rename(columns={'approx_cost(for two people)': 'cost'})
```

```python
df['online_order'] = df['online_order'].map({'Yes': 1, 'No': 0})
df['book_table'] = df['book_table'].map({'Yes': 1, 'No': 0})
```

```python
df = df.dropna(subset=['rate', 'cost', 'cuisines', 'location'])
```

```python
df['location'] = df['location'].str.strip().str.title()
df['listed_in(city)'] = df['listed_in(city)'].str.strip().str.title()
```

```python
df_cuisine = df.assign(cuisine=df['cuisines'].str.split(', ')).explode('cuisine')
df_cuisine['cuisine'] = df_cuisine['cuisine'].str.strip()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Boxplot se visually dekho
plt.figure(figsize=(8,4))
sns.boxplot(x=df['votes'])
plt.title('Votes - Outlier Check (Boxplot)')
plt.show()

# 2. IQR method se numerically calculate karo
Q1 = df['votes'].quantile(0.25)
Q3 = df['votes'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['votes'] < lower_bound) | (df['votes'] > upper_bound)]
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Total outliers:", outliers.shape[0])
print("Percentage of data:", round(outliers.shape[0]/df.shape[0]*100, 2), "%")
```

```python
import numpy as np

df['votes_log'] = np.log1p(df['votes'])
plt.figure(figsize=(8,4))
sns.histplot(df['votes_log'], bins=30)
plt.title('Log-transformed Votes Distribution')
plt.show()     
```
